# ITERATOR

## What is an Iterator?
An iterator is an object that lets us go through a collection one item at a time.

**Example**

```python
numbers = [10, 20, 30, 40]

for num in numbers:
    print(num)
```

## Internal Work
This loop does look simple but python is actually doing this in background for each iteration:
1. Create an iterator
2. Ask it for the next vakue
3. Print it
4. Ask again and repeat it until no values are left

## Create an Iterator

Use `iter()` to create an iterator

```python
nums = [1, 2, 3, 4, 5]
it = iter(nums)

print(type(it))
print(it)
```

Now it is an iterator.  
Use `next()` to get values manually

```python
print(next(it)) # moves to first element
print(next(it)) # moves to 2nd element
print(next(it)) # moves to 3rd element
print(next(it)) # moves to 4th element

# if we use it one more time it will go out of bound
```

## Why doesn't a `for` loop crash?
Because for handles StopIteration automatically.

## Iterable vs Iterator

**Iterable:** An object we can create an iterator from.  
Examples: `list`, `tuple`, `string`, `set`, `dictionary`, `file`, `range`  

**Iterator:** The object returned by iter().  
`it = iter(numbers)` -> here it is an iterator

## Iterators vs Lists

| List                | Iterator                       |
| ------------------- | ------------------------------ |
| Stores all elements | Produces one element at a time |
| Can access by index | No indexing                    |
| Can restart easily  | Gets exhausted                 |
| Uses more memory    | Uses less memory               |
| Reusable            | Usually single-use             |


```python
nums = [10, 20, 30]

it = iter(nums) 

print(next(it))  # moved to 10 and print 10

for x in it: 
    print(x) # 20 30

print(list(it)) # empty list
```
