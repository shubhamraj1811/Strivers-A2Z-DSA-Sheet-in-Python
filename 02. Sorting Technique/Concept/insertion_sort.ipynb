# INSERTION SORT

## 1️⃣ What is Insertion Sort?
> Insertion Sort builds a sorted array one element at a time by picking each element and inserting it into its correct position in the already sorted part.

---

## Real Life Analogy 🃏

Think of how you sort playing cards in your hand:

- You pick one card at a time
- You compare it with the cards already in your hand
- You slide it into the right position
- Repeat until all cards are sorted

**Example — Sort [5, 3, 8, 1, 2]**

```
Initial Array → [5, 3, 8, 1, 2]

Pass 1 → Pick 3
         [5, 3, 8, 1, 2]
          ↑  ↑
          3 < 5 → shift 5 right
         [3, 5, 8, 1, 2]  ✅

Pass 2 → Pick 8
         [3, 5, 8, 1, 2]
             ↑  ↑
               8 > 5 → already in place
         [3, 5, 8, 1, 2]  ✅

Pass 3 → Pick 1
         [3, 5, 8, 1, 2]
          ↑  ↑  ↑  ↑
          1 < 8 → shift 8
          1 < 5 → shift 5
          1 < 3 → shift 3
         [1, 3, 5, 8, 2]  ✅

Pass 4 → Pick 2
         [1, 3, 5, 8, 2]
             ↑  ↑  ↑  ↑
          2 < 8 → shift 8
          2 < 5 → shift 5
          2 < 3 → shift 3
          2 > 1 → STOP, insert here
         [1, 2, 3, 5, 8]  ✅

Final Sorted Array → [1, 2, 3, 5, 8] 🎉
```

## 2️⃣ ALL POSSIBLE APPROACHES
---

### 🔴 Approach 1 — Using an Extra Array (Naive)

1. Create a new empty result array
2. For each element in the original array
3. Find the correct position in the result array
4. Insert it there by shifting elements

```
TC → O(n²)
SC → O(n)  ← extra array used
```

**❌ Wastes extra space — never used in practice**

---

### 🟡 Approach 2 — Using Binary Search (Binary Insertion Sort)

1. For each element, use binary search to find its correct position in the sorted part
2. Shift elements right to make room
3. Insert the element

```
TC → O(n log n) for comparisons
     O(n²) overall — shifting still takes O(n)
SC → O(1)
```

> ⚠️ Reduces comparisons but shifting cost still makes it O(n²) overall

---

### 🟢 Approach 3 — Standard Insertion Sort ⭐ OPTIMAL (for this algorithm)

1. Start from index 1 (first element is already "sorted")
2. Pick the current element as key
3. Compare key with elements to its left
4. Shift all elements greater than key one position to the right
5. Place key in its correct position
6. Repeat for all elements

```
TC → O(n²) worst & average case
     O(n)  best case  ← already sorted array!
SC → O(1)  ← in-place sorting ⭐
```

> At every step, the left portion is always sorted.
> We just need to find the right spot for the current element and slide it in.

## 3️⃣ STEP-BY-STEP ALGORITHM 📋

```
1. Start from index i = 1 (element at index 0 is already sorted)

2. For every i from 1 to n-1:
      a. Store arr[i] as key  ← the element we want to insert
      b. Set j = i - 1       ← pointer to the sorted portion

3. Inner while loop — shift elements right:
      While j >= 0 AND arr[j] > key:
            arr[j + 1] = arr[j]   ← shift element one position right
            j = j - 1             ← move pointer left

4. Place key in its correct position:
      arr[j + 1] = key

5. Repeat until i reaches the end of array

6. Array is now sorted in-place
```

## 4️⃣ CODE IMPLEMENTATION

Array = [8, 3, 5, 1, 4]

````python
def insertionSort(ar):
    # store the length
    n = len(ar)

    for i in range(1,n): # outer loop
        current = ar[i] # pick current element/Key
        j = i-1 # sorted portion

        while j >= 0 and ar[j] > current:
            ar[j+1] = ar[j] # shift j right
            j -= 1 # decre j

        ar[j+1] = current # insert key/current at correct position

    return ar

# example
array = [8, 3, 5, 1, 4]
print(f"Unsorted Array: {array}")
insertionSort(array)
print(f"Sorted Array: {array}")
````
