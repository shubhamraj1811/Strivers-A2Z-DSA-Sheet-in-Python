# Merge Sort 🔀

## 1️⃣ What is Merge Sort?

Merge Sort is a Divide and Conquer algorithm that splits the array into halves, recursively sorts each half, and then merges them back together in sorted order.

**Example —** Sort [38, 27, 43, 3, 9, 82, 10]  

```
DIVIDE PHASE 🔽
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

      [38, 27, 43, 3, 9, 82, 10]
          ↙              ↘
  [38, 27, 43]        [3, 9, 82, 10]
     ↙     ↘            ↙         ↘
[38, 27]  [43]      [3, 9]     [82, 10]
  ↙    ↘               ↙    ↘      ↙    ↘
[38]  [27]           [3]   [9]  [82]  [10]


MERGE PHASE 🔼
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

  [38]  [27]          [3]   [9]        [82]  [10]
  ↘    ↙               ↘    ↙          ↘    ↙
[27, 38]     [43]       [3, 9]         [10, 82]
     ↘     ↙               ↘         ↙
  [27, 38, 43]            [3, 9, 10, 82]
          ↘              ↙
  [3, 9, 10, 27, 38, 43, 82]  ✅
```

## 2️⃣ ALL POSSIBLE APPROACHES

---

### 🔴 Approach 1 — Brute Force (Nested Loops / Insertion Sort style)

1. Pick each element one by one
2. Compare it with all others
3. Place it in the correct position by shifting

```
TC → O(n²)  worst and average case
SC → O(1)
```

> ❌ Too slow for large arrays — no divide and conquer benefit

---

### 🟡 Approach 2 — Top Down Merge Sort (Recursive) ⭐ CLASSIC

1. Find the middle of the array
2. Recursively sort the left half
3. Recursively sort the right half
4. Merge both sorted halves into one sorted array

```
TC → O(n log n)  all cases (best, avg, worst)
SC → O(n)        extra space for temp arrays during merge
                 O(log n) recursive call stack
```

> ✅ Clean, intuitive, most commonly used

---

### 🟢 Approach 3 — Bottom Up Merge Sort (Iterative)

1. Start by treating each element as a sorted subarray of size 1
2. Merge adjacent subarrays of size 1 → sorted subarrays of size 2
3. Merge adjacent subarrays of size 2 → sorted subarrays of size 4
4. Keep doubling the size until the entire array is merged

```
TC → O(n log n)  all cases
SC → O(n)        no recursion stack needed
```

> ✅ No recursion — avoids stack overflow for very large arrays

## 3️⃣ OPTIMAL APPROACH — Top Down Merge Sort Deep Dive 🔍

> Merging two already-sorted arrays is easy and fast — O(n)
> So if we keep dividing until we hit single elements (which are always sorted),
> we can merge our way back up to a fully sorted array!

**Time Complexity:**
```
log n  → number of times we divide the array in half
n      → work done at each level to merge everything

Total  → n × log n = O(n log n) 🔥
```

### Dry Run

**Eample: [5, 2, 8, 1]**

```
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
DIVIDE
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

            [5, 2, 8, 1]
              mid = 1
           ↙         ↘
       [5, 2]         [8, 1]
        mid=0          mid=0
      ↙    ↘          ↙    ↘
    [5]    [2]       [8]    [1]


━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
MERGE
━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━

Merge [5] and [2]:
  compare 5 vs 2 → take 2
  leftover → take 5
  result → [2, 5] ✅

Merge [8] and [1]:
  compare 8 vs 1 → take 1
  leftover → take 8
  result → [1, 8] ✅

Merge [2, 5] and [1, 8]:
  compare 2 vs 1 → take 1  → [1]
  compare 2 vs 8 → take 2  → [1, 2]
  compare 5 vs 8 → take 5  → [1, 2, 5]
  leftover → take 8         → [1, 2, 5, 8] ✅

Final → [1, 2, 5, 8] 🎉
```

### STEP-BY-STEP ALGORITHM 📋

MAIN FUNCTION — mergeSort(arr):

1. If array has 0 or 1 element → return (base case, already sorted)
2. Find mid = length // 2
3. Split into left  = arr[0 : mid]
              right = arr[mid : end]
4. Recursively call mergeSort(left)
5. Recursively call mergeSort(right)
6. Call merge(arr, left, right) to combine them


MERGE FUNCTION — merge(arr, left, right):

1. Initialize three pointers:
      i = 0  → pointer for left array
      j = 0  → pointer for right array
      k = 0  → pointer for main array

2. While i < len(left) AND j < len(right):
      If left[i] <= right[j]:
            arr[k] = left[i]
            i++
      Else:
            arr[k] = right[j]
            j++
      k++

3. Copy remaining elements of left (if any):
      While i < len(left):
            arr[k] = left[i]
            i++, k++

4. Copy remaining elements of right (if any):
      While j < len(right):
            arr[k] = right[j]
            j++, k++

### PSEUDOCODE 📝

```
function mergeSort(arr):
    if length(arr) <= 1:
        return arr                       ← base case

    mid   = length(arr) // 2
    left  = arr[0 : mid]                ← left half
    right = arr[mid : end]              ← right half

    mergeSort(left)                     ← sort left recursively
    mergeSort(right)                    ← sort right recursively

    merge(arr, left, right)             ← merge both halves


function merge(arr, left, right):
    i = 0, j = 0, k = 0

    while i < len(left) AND j < len(right):
        if left[i] <= right[j]:
            arr[k] = left[i]
            i++
        else:
            arr[k] = right[j]
            j++
        k++

    while i < len(left):               ← copy leftover left elements
        arr[k] = left[i]
        i++, k++

    while j < len(right):              ← copy leftover right elements
        arr[k] = right[j]
        j++, k++
```
