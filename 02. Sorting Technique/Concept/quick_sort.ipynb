# #️⃣ QUICK SORT

Quick Sort is a Divide and Conquer sorting algorithm.  
Instead of repeatedly moving the largest element (like Bubble Sort), Quick Sort:  
1. Picks one element called the pivot.
2. Rearranges the array so:
   - smaller elements go to the left
   - larger elements go to the right
3. Repeats the same process on the left and right parts.

---

## ❇️ ANALOGY

Imagine you have to sort 1,000 playing cards.  
Would you compare every card with every other card?  

You'd first split them into groups:  
- small cards
- big cards

Then sort each group separately.  

## ❇️ WHY QUICK SORT

- It is very fast on average.
- It works in-place (for common implementations).
- It is one of the most frequently asked sorting algorithms in interviews.

### 🛡️ Example

Array : [7, 2, 9, 4, 6]  
Suppose the pivot is: 6  
After partitioning: [2, 4, 6, 7, 9]  

- Every element left of 6 is smaller than 6.
- Every element right of 6 is greater than 6.
- 6 is now in its final sorted position.

This is the key idea of Quick Sort.  

---

## ❇️ PIVOT

The pivot is the element we choose as the middle reference for one partition step.  
We use it to split the array into two groups:

- elements smaller than pivot
- elements greater than pivot

Think of the pivot like a leader in a line.  
Everyone else is compared to that leader: 

- shorter people go left
- taller people go right

The leader is then placed in the correct final position.  

---

## ❇️ DRY RUN

`Array = [8, 3, 1, 7, 0, 10, 2]`

**STEP 1:**

Pivot = 2
Now compare every element with 2:  

- 8 > 2 → right side
- 3 > 2 → right side
- 1 < 2 → left side
- 7 > 2 → right side
- 0 < 2 → left side
- 10 > 2 → right side

After Partition : [1, 0, 2, 7, 8, 10, 3]  

**STEP 2:**
Sort the left part [1, 0]  
Sort the right part [7, 8, 10, 3]
