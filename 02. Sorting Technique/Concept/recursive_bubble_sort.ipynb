# #️⃣ RECURSIVE BUBBLE SORT

❇️ ## WHAT IS RECURSIVE BUBBLE SORT?
- Recursive Bubble Sort is just Bubble Sort written using recursion instead of a loop.
- In the recursive version, after one full pass is done, the function calls itself again on the smaller unsorted part.


---

### 🔰 Example
`Array = [5, 1, 8, 4]`  
**First Pass:**  
- compare 5 and 1 → swap → [1, 5, 4, 2]
- compare 5 and 4 → swap → [1, 4, 5, 2]
- compare 5 and 2 → swap → [1, 4, 2, 5]

Now the largest element 5 is fixed at the end.  
- now sort the left part [1, 4, 2]
- then sort [1, 4]
- then sort [1]
